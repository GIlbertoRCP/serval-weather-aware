```python
import folium
import os
import re
import xml.etree.ElementTree as ET
import pickle
import pandas as pd
```

```python
image_labels=pickle.load(open('../results/final_result.pkl','rb'))
image_labels={k:(v['forest'][0]>0.1)*1 for k,v in image_labels.items()}
```

```python
all_metadata=open('../data/all_metadata','r').readlines()
# randomly choose 1000 images
all_metadata=all_metadata[::10]
```

```python
metadata=[]
for filename in all_metadata:
    filename=filename.strip()
    if filename.endswith('.xml'):
        tree = ET.parse(filename)
        root = tree.getroot()
        # Extract the center position from the XML and convert it to a list of floats
        center_pos = list(map(float, root.find('.//{http://www.opengis.net/gml}pos').text.split()))
        # Extract the image id from the filename using regular expressions
        match = re.search(r'(\w+)_3B_AnalyticMS_metadata.xml', filename)
        # if match:
        image_id = match.group(1)
        # Add the image id, center latitude, and center longitude to the metadata list
        metadata.append({
            'id': image_id,
            'latitude': center_pos[1],
            'longitude': center_pos[0]
        })
```

```python
pd.DataFrame(metadata)
```

```python

# Merge the image labels with the metadata
merged_data = pd.DataFrame(metadata).merge(pd.DataFrame(image_labels.items(), columns=['id', 'label']), left_on='id', right_on='id')

# Create a folium map centered on the first coordinate in the metadata
m = folium.Map(location=[merged_data['latitude'].iloc[0], merged_data['longitude'].iloc[0]], zoom_start=10)

# Iterate over each row in the merged dataframe and add a marker to the map
for index, row in merged_data.iterrows():
    # Choose a color based on the label
    color = 'green' if row['label'] == 0 else 'red'
    # Add a marker with a popup displaying the image id and label
    folium.Marker([row['latitude'], row['longitude']], popup=f"{row['id']}: {row['label']}", icon=folium.Icon(color=color)).add_to(m)

# Display the map
m
```
